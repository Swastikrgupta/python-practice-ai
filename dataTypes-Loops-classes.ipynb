```python
from pydantic import BaseModel
```

```python
class Car(BaseModel):
    model: str
    serial_no: int
    
''' 
pydantic classes enforces data types
'''
```

```python
 car_obj = Car(model="acs", serial_no=35453)
```

```python
car_obj.model
```

```python
class YoutubeChannel(BaseModel):
    channel_name: str
    sub_count: int = 0
    def __innit__(self, name):
        self.channel_name = name
    def subscribe(self):
        self.sub_count+=1
        print(f'Thanks for subscribing! Total: {self.sub_count}')
    def unsubscribe(self):
        self.sub_count= max(0, self.sub_count-1)
    def show(self):
        print(f'{self.channel_name} has {self.sub_count} subs')
    
```

```python
codebasics = YoutubeChannel(channel_name="codebasics")
```

```python
codebasics.subscribe()
```

```python
codebasics.show()
```

```python
codebasics.unsubscribe()
```
