```python
conversations = []
```

```python
def chat(user_query):
    conversations.append({'type': 'user_input', 'content': [{'type': 'text' , 'text': user_query}]})
    interaction = client.interactions.create(
        model="gemini-3.8-flash",
        input=conversations
    )
    conversations.append({'type': 'model_output', 'content': [{'type': 'text' , 'text': interaction.output_text}]})
    print(interaction.output_text)
    
```

```python
chat('Hi, i am Swastik.')
```

```python
chat('Whats my name?')
```

```python
for item in conversations:
    print(f'item: {item['type']} -> {item['content']}\n')
```

```python
from pydantic import BaseModel
```

```python
user_review= "I ordered the wireless earbuds last week. Sound quality is amazing and battery lasts all day, but the case feels cheap and it arrived two days late. Probably won't buy from this brand again."

class ReviewAnalysis(BaseModel):
    sentiment: str
    rating: int
    pro: list[str]
    cons: list[str]

def ask_gemini():
    interaction = client.interactions.create(
        model="gemini-3.5-flash-lite",
        input=user_review,
        system_instruction="Analyze the customer review.",
        response_format={
            'type': 'text',
            'mime_type': 'application/json',
            'schema': ReviewAnalysis.model_json_schema()
        }
    )
    return ReviewAnalysis.model_validate_json(interaction.output_text)
    
```

```python
result = ask_gemini()
print(result)
for con, pros in zip(result.cons , result.pro):
    print(f'con, pros: {con}, {pros}')
```
