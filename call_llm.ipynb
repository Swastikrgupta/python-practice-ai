```python
from dotenv import load_dotenv
```

```python
from google import genai

load_dotenv()
client = genai.Client()

interaction = client.interactions.create(
    model="gemini-3.8-flash",
    input="Tell me who invented Samosa in 2-3 liens"
)

print(interaction.output_text)
```

```python
ingradients = ['spinach','paneer','onions', 'ghee', "masala's"]
','.join(ingradients)

cuisine = 'Indian'
diet = 'Vegetarian'
```

```python
prompt = f'''
Generate one food receipe using these ingradients: {','.join(ingradients)}
Receipe should not be more than 30 words
Cuisine: {cuisine}
Diet: {diet}
'''
```

```python
interaction = client.interactions.create (
    model = "gemini-3.8-flash",
    input = prompt
)

print(interaction.output_text)
```
