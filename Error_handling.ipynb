```python
def total_chai_price(*args):
    try:
        return int(args[0])*int(args[1]) if int(args[0])>0 else print('Cups should be atleast 1')
    except Exception as e:
        print(f'Enter numbers only: {e}')
total_chai_price(2,3)
```

```python
total_chai_price('2','3')
```

```python
total_chai_price("two")
```

```python
total_chai_price(1.00,2.2)
```
